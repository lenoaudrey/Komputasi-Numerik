```python
# ===== CELL 1: FUNGSI DAN PARAMETER =====
# Metode Bisection mencari akar persamaan non linear f(x) = 0
# dengan cara membagi dua selang [a, b] berulang kali.

# Mengimpor pandas untuk menampilkan tabel iterasi
import pandas as pd

# Mendefinisikan fungsi yang akan dicari akarnya
# Contoh: f(x) = x^3 - x - 2   (ganti sesuai soal)
def f(x):
    return x**3 - x - 2

a = 1.0            # batas bawah selang awal
b = 2.0            # batas atas selang awal
tol = 1e-6         # toleransi error (1e-6 = 0,000001)
maks_iter = 100    # batas iterasi agar perulangan tidak berjalan tanpa akhir
```

```python
# ===== CELL 2: MEMERIKSA SYARAT METODE BISECTION =====
# Akar dijamin ada di [a, b] jika f(a) dan f(b) berbeda tanda,
# artinya f(a) * f(b) bernilai negatif.

# Menampilkan nilai fungsi di kedua ujung selang
print("f(a) =", f(a))
print("f(b) =", f(b))

# Jika hasil kali tidak negatif, tanda sama, akar tidak terjamin ada
if f(a) * f(b) >= 0:
    raise ValueError("f(a) dan f(b) harus berbeda tanda.")
```

```python
# ===== CELL 3: PROSES ITERASI BISECTION =====

# List kosong untuk menyimpan data setiap iterasi (untuk ditampilkan sebagai tabel)
tabel = []

# Perulangan dari iterasi ke-1 sampai iterasi maksimum
for i in range(1, maks_iter + 1):
    # Menghitung titik tengah selang
    c = (a + b) / 2

    # Menyimpan nomor iterasi, a, b, c, dan f(c) ke dalam tabel
    tabel.append([i, a, b, c, f(c)])

    # Kriteria berhenti: f(c) sudah cukup dekat dengan nol,
    # atau selang sudah cukup kecil. Jika ya, keluar dari perulangan
    if abs(f(c)) < tol or (b - a) / 2 < tol:
        break

    # Menentukan setengah selang mana yang masih memuat akar
    if f(a) * f(c) < 0:
        b = c      # akar ada di [a, c], jadi batas atas dipindah ke c
    else:
        a = c      # akar ada di [c, b], jadi batas bawah dipindah ke c
```

```python
# ===== CELL 4: MENAMPILKAN TABEL ITERASI DAN HASIL =====

# Mengubah list tabel menjadi DataFrame agar tampil rapi
df = pd.DataFrame(tabel, columns=["Iterasi", "a", "b", "c", "f(c)"])
print(df)

# Menampilkan akar hampiran (c terakhir), nilai f di akar, dan jumlah iterasi
# {c:.6f} = 6 angka di belakang koma, {f(c):.2e} = notasi ilmiah
print(f"\nAkar hampiran = {c:.6f}")
print(f"f(akar)       = {f(c):.2e}")
print(f"Jumlah iterasi = {i}")
```

```python
# ===== CELL 5: VERIFIKASI HASIL =====
# Jika akar benar, f(akar) harus sangat dekat dengan nol (lebih kecil dari toleransi)
print("Nilai f(akar) mendekati nol?", abs(f(c)) < tol)
```
