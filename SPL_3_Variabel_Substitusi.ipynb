```python
a1, b1, c1, d1 = 2, 1, -1, 8
a2, b2, c2, d2 = -3, -1, 2, -11
a3, b3, c3, d3 = -2, 1, 2, -3

print(f"({a1})x + ({b1})y + ({c1})z = {d1}")
print(f"({a2})x + ({b2})y + ({c2})z = {d2}")
print(f"({a3})x + ({b3})y + ({c3})z = {d3}")
```

```python
if a1 == 0:
    if a2 != 0:
        a1, b1, c1, d1, a2, b2, c2, d2 = a2, b2, c2, d2, a1, b1, c1, d1
    elif a3 != 0:
        a1, b1, c1, d1, a3, b3, c3, d3 = a3, b3, c3, d3, a1, b1, c1, d1
    else:
        raise ValueError("Koefisien x semuanya nol.")

print("a1 =", a1)
```

```python
p1 = b2 - a2 * b1 / a1
q1 = c2 - a2 * c1 / a1
r1 = d2 - a2 * d1 / a1

p2 = b3 - a3 * b1 / a1
q2 = c3 - a3 * c1 / a1
r2 = d3 - a3 * d1 / a1

print(f"({p1})y + ({q1})z = {r1}")
print(f"({p2})y + ({q2})z = {r2}")
```

```python
if p1 == 0:
    p1, q1, r1, p2, q2, r2 = p2, q2, r2, p1, q1, r1

if p1 == 0:
    raise ValueError("Tidak ada solusi tunggal.")
```

```python
penyebut = q2 - p2 * q1 / p1

if penyebut == 0:
    raise ValueError("Tidak ada solusi tunggal.")

z = (r2 - p2 * r1 / p1) / penyebut
y = (r1 - q1 * z) / p1
x = (d1 - b1 * y - c1 * z) / a1
```

```python
print(f"x = {x:.4f}")
print(f"y = {y:.4f}")
print(f"z = {z:.4f}")
```

```python
print(a1*x + b1*y + c1*z, "=", d1)
print(a2*x + b2*y + c2*z, "=", d2)
print(a3*x + b3*y + c3*z, "=", d3)
```
